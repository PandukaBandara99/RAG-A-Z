
# **Introduction to Graph Databases**

## **What is a Graph Database?**

A **Graph Database** is a type of database that stores data using a **graph structure** —
made up of **nodes (entities)** and **relationships (connections)**.

Unlike traditional databases that use **tables and rows**, graph databases represent data the way we naturally think: as connected objects.

**Example:**

```
(Alice)-[:FRIENDS_WITH]->(Bob)
```

This means “Alice is friends with Bob”.

### Key Components

* **Node** → represents an entity (e.g., a person, city, or product)
* **Relationship** → connects nodes (e.g., FRIENDS_WITH, LIVES_IN)
* **Property** → adds details as key–value pairs (e.g., `{age: 30}`)

In short
**Graph Databases = Data + Relationships + Context**

---

## **The Evolution of Data Storage**

| Era         | Type                     | Description                                         | Example            |
| ----------- | ------------------------ | --------------------------------------------------- | ------------------ |
| 1970s–1990s | **Relational Databases** | Store data in tables with fixed schemas.            | MySQL, PostgreSQL  |
| 2000s–2010s | **NoSQL Databases**      | Handle large-scale and unstructured data.           | MongoDB, Cassandra |
| 2010s–Now   | **Graph Databases**      | Designed for connected and relationship-heavy data. | Neo4j, ArangoDB    |

➡️ As data became **more connected** (e.g., social networks, recommendations),
**Graph Databases** became essential.

---

## **Relational vs NoSQL vs Graph**

| Feature            | Relational        | NoSQL                   | Graph                 |
| ------------------ | ----------------- | ----------------------- | --------------------- |
| **Model**          | Tables            | Documents / Key-Value   | Nodes & Relationships |
| **Schema**         | Fixed             | Flexible                | Flexible              |
| **Relationships**  | JOINs (expensive) | Weak or manual          | Native & Fast         |
| **Use Case**       | Structured data   | Big / Unstructured data | Connected data        |
| **Query Language** | SQL               | Mongo Query, etc.       | Cypher                |

**Example Query:**

* In SQL → Multiple JOINs to find “friends of friends”
* In Graph → Simple traversal:

  ```
  MATCH (a)-[:FRIENDS_WITH*2]->(b)
  RETURN b
  ```

## Core Concepts of Graph Theory

### Nodes

* Represent entities (e.g., `Person`, `Movie`, `Product`)
* Example:
  `(:Person {name: "Alice", age: 30})`

### Relationships

* Represent how nodes are connected.
  Example:
  `(Alice)-[:WORKS_AT {since: 2020}]->(Neo4j_Inc)`

### Directed vs Undirected

* **Directed:** `(A)-[:LIKES]->(B)`
* **Undirected:** `(A)--(B)`

> Neo4j uses **directed relationships** by default.

### Labels & Properties

* **Labels:** classify nodes (like a “type”)
* **Properties:** key–value pairs attached to nodes/relationships

  ```
  (:Person {name: "Bob", age: 25})
  [:FRIENDS_WITH {since: 2018}]
  ```

## Basic Setup & Connection Example

### Python Example:

```python
from neo4j import GraphDatabase

uri = "bolt://localhost:7687"
driver = GraphDatabase.driver(uri, auth=("neo4j", "password"))

def get_friends(tx, name):
    query = """
    MATCH (p:Person {name: $name})-[:FRIENDS_WITH]->(friend)
    RETURN friend.name
    """
    result = tx.run(query, name=name)
    return [record["friend.name"] for record in result]

with driver.session() as session:
    print(session.execute_read(get_friends, name="Alice"))
```

#  **Introduction to Cypher**

---
## **What is Cypher?**

**Cypher** is the **query language for Neo4j**, specifically designed to work with graph data.
It is **declarative**, which means you **describe what you want**, not how to get it — similar to SQL but for graphs.

### Key Features

* **Intuitive** syntax for graphs
* Works with **nodes, relationships, and properties**
* **Pattern matching** to traverse connections
* Easy to **create, read, update, and delete** graph data

# **CRUD Operations in Cypher**

CRUD in Neo4j using Cypher includes **Create, Read, Update, and Delete** operations. 

---

## **1. CREATE – Adding Nodes and Relationships**

The `CREATE` clause adds new nodes or relationships to the graph.

### Example: Creating Nodes

**Query:**

```cypher
CREATE (alice:Person {name: "Alice", age: 30})
CREATE (bob:Person {name: "Bob", age: 28})
```

**Output in Neo4j Browser:**

* Two nodes are created:

  * `Alice:Person {name:"Alice", age:30}`
  * `Bob:Person {name:"Bob", age:28}`

**Visual:**

```
(Alice:Person)   (Bob:Person)
```

---

### Example: Creating a Relationship

**Query:**

```cypher
MATCH (a:Person {name: "Alice"}), (b:Person {name: "Bob"})
CREATE (a)-[:FRIENDS_WITH {since: 2020}]->(b)
```

**Output:**

* A relationship is created:

```
(Alice)-[:FRIENDS_WITH {since: 2020}]->(Bob)
```

---

## **2. MATCH – Reading Data**

`MATCH` is used to **find nodes and relationships**.

### Example: Find All Persons

**Query:**

```cypher
MATCH (p:Person)
RETURN p.name, p.age
```

**Output Table:**

| p.name | p.age |
| ------ | ----- |
| Alice  | 30    |
| Bob    | 28    |

### Example: Find Friends of Alice

**Query:**

```cypher
MATCH (a:Person {name: "Alice"})-[:FRIENDS_WITH]->(friend)
RETURN friend.name
```

**Output:**

| friend.name |
| ----------- |
| Bob         |

---

## **3. MERGE – Find or Create**

`MERGE` ensures a node or relationship exists, creating it only if it doesn’t exist.

### Example: Merge Node

**Query:**

```cypher
MERGE (charlie:Person {name: "Charlie"})
RETURN charlie.name
```

**Output:**

| charlie.name |
| ------------ |
| Charlie      |

### Example: Merge Relationship

**Query:**

```cypher
MATCH (a:Person {name:"Alice"}), (c:Person {name:"Charlie"})
MERGE (a)-[:FRIENDS_WITH]->(c)
RETURN a.name, c.name
```

**Output Table:**

| a.name | c.name  |
| ------ | ------- |
| Alice  | Charlie |

**Key Difference from CREATE:**

* `MERGE` won’t create duplicates if the node or relationship already exists.

---

## **4. DELETE – Removing Nodes and Relationships**

`DELETE` removes nodes or relationships.

### Example: Delete Relationship

**Query:**

```cypher
MATCH (a:Person {name:"Alice"})-[r:FRIENDS_WITH]->(b:Person {name:"Bob"})
DELETE r
```

**Output:**

* The `FRIENDS_WITH` relationship between Alice and Bob is removed.
* Nodes `Alice` and `Bob` remain.

### Example: Delete Node

**Query:**

```cypher
MATCH (n:Person {name:"Bob"})
DETACH DELETE n
```

**Output:**

* Node `Bob` and all relationships attached to Bob are deleted.
* Remaining nodes: Alice, Charlie

---

## **5. SET – Updating Node and Relationship Properties**

`SET` updates or adds properties.

### Example: Update Node Properties

**Query:**

```cypher
MATCH (alice:Person {name:"Alice"})
SET alice.age = 31, alice.city = "Colombo"
RETURN alice.name, alice.age, alice.city
```

**Output:**

| alice.name | alice.age | alice.city |
| ---------- | --------- | ---------- |
| Alice      | 31        | Colombo    |

### Example: Update Relationship Properties

**Query:**

```cypher
MATCH (a:Person {name:"Alice"})-[r:FRIENDS_WITH]->(c:Person {name:"Charlie"})
SET r.since = 2021
RETURN a.name, r.since, c.name
```

**Output:**

| a.name | r.since | c.name  |
| ------ | ------- | ------- |
| Alice  | 2021    | Charlie |

### Example: Remove a Property

**Query:**

```cypher
MATCH (alice:Person {name:"Alice"})
REMOVE alice.city
RETURN alice.name, alice.age, alice.city
```

**Output:**

| alice.name | alice.age | alice.city |
| ---------- | --------- | ---------- |
| Alice      | 31        | null       |

# **Filtering and Pattern Matching in Cypher**

Graph queries become powerful when you **filter nodes, relationships, and patterns** to get exactly what you need. Cypher provides **flexible filtering and pattern-matching capabilities**.

---

## **1. Using WHERE – Filtering Data**

The `WHERE` clause is used to **filter nodes and relationships** based on properties.

### Example: Simple Filtering

**Query:**

```cypher
MATCH (p:Person)
WHERE p.age > 25
RETURN p.name, p.age
```

**Output:**

| p.name | p.age |
| ------ | ----- |
| Alice  | 30    |
| Bob    | 28    |

**Explanation:** Only people older than 25 are returned.

---

### Using Comparison Operators

* `=` → equals
* `<>` → not equals
* `<`, `>`, `<=`, `>=` → less/greater than
* `AND`, `OR`, `NOT` → combine conditions

**Example:**

```cypher
MATCH (p:Person)
WHERE p.age > 25 AND p.city = "Colombo"
RETURN p.name, p.age
```

---

## **2. Using IN – Matching Multiple Values**

`IN` checks if a property matches any value in a **list**.

**Query:**

```cypher
MATCH (p:Person)
WHERE p.name IN ["Alice", "Charlie"]
RETURN p.name
```

**Output:**

| p.name  |
| ------- |
| Alice   |
| Charlie |

---

## **3. Using String Functions – CONTAINS, STARTS WITH, ENDS WITH**

### CONTAINS – Partial Match

**Query:**

```cypher
MATCH (p:Person)
WHERE p.name CONTAINS "li"
RETURN p.name
```

**Output:**

| p.name  |
| ------- |
| Alice   |
| Charlie |

**Explanation:** Returns nodes where `name` contains “li”.

---

### STARTS WITH – Beginning of String

**Query:**

```cypher
MATCH (p:Person)
WHERE p.name STARTS WITH "A"
RETURN p.name
```

**Output:**

| p.name |
| ------ |
| Alice  |

---

### ENDS WITH – Ending of String

**Query:**

```cypher
MATCH (p:Person)
WHERE p.name ENDS WITH "e"
RETURN p.name
```

**Output:**

| p.name |
| ------ |
| Alice  |

---

## **4. Matching Patterns and Subgraphs**

Graph pattern matching allows you to **find connected nodes** or **subgraphs**.

### Example: Direct Relationship

```cypher
MATCH (a:Person)-[:FRIENDS_WITH]->(b:Person)
RETURN a.name, b.name
```

**Output:**

| a.name | b.name  |
| ------ | ------- |
| Alice  | Bob     |
| Alice  | Charlie |

---

### Example: Subgraph / Multiple Hops

```cypher
MATCH (a:Person)-[:FRIENDS_WITH*2]->(c:Person)
RETURN a.name, c.name
```

**Explanation:**

* `*2` → traverses 2 relationships
* Finds friends of friends

**Output:**

| a.name | c.name  |
| ------ | ------- |
| Alice  | Charlie |

---

### Example: Optional Pattern (MAY Exist)

```cypher
MATCH (p:Person)
OPTIONAL MATCH (p)-[:FRIENDS_WITH]->(friend)
RETURN p.name, friend.name
```

**Explanation:** Returns all persons, including those without friends (friend.name = null).

---

## **5. Combining Conditions**

You can **combine multiple filters** using `AND`, `OR`, and parentheses for clarity.

**Query:**

```cypher
MATCH (p:Person)-[:FRIENDS_WITH]->(friend)
WHERE p.age > 25 AND (friend.age < 30 OR friend.city = "Colombo")
RETURN p.name, friend.name, friend.age, friend.city
```

**Output Table:**

| p.name | friend.name | friend.age | friend.city |
| ------ | ----------- | ---------- | ----------- |
| Alice  | Bob         | 28         | Kandy       |
| Alice  | Charlie     | 26         | Colombo     |

**Explanation:**

* Only friends matching the **age < 30 OR city = Colombo** condition are returned.
* Logical operators allow **complex filtering** in graph queries.

---

## **6. Practical Example: Filtering & Pattern Matching Workflow**

**Step 1: Create Sample Data**

```cypher
CREATE (alice:Person {name:"Alice", age:30, city:"Colombo"})
CREATE (bob:Person {name:"Bob", age:28, city:"Kandy"})
CREATE (charlie:Person {name:"Charlie", age:26, city:"Colombo"})
CREATE (alice)-[:FRIENDS_WITH]->(bob)
CREATE (alice)-[:FRIENDS_WITH]->(charlie)
```

**Step 2: Filter Friends in Colombo**

```cypher
MATCH (alice:Person {name:"Alice"})-[:FRIENDS_WITH]->(friend)
WHERE friend.city = "Colombo"
RETURN friend.name, friend.city
```

**Output:**

| friend.name | friend.city |
| ----------- | ----------- |
| Charlie     | Colombo     |

**Step 3: Filter by Age**

```cypher
MATCH (alice:Person {name:"Alice"})-[:FRIENDS_WITH]->(friend)
WHERE friend.age < 30
RETURN friend.name, friend.age
```

**Output:**

| friend.name | friend.age |
| ----------- | ---------- |
| Bob         | 28         |
| Charlie     | 26         |

# **Aggregation and Ordering in Cypher**

Aggregation allows you to **summarize or group data** in a graph, while ordering lets you **control the sequence of results**. These operations are essential for **analysis and reporting**.

---

## **1. RETURN – Displaying Data**

The `RETURN` clause is used to **display results** of a query.
It can return **raw properties, calculated values, or aggregates**.

**Example: Return names and ages**

```cypher
MATCH (p:Person)
RETURN p.name, p.age
```

**Output:**

| p.name  | p.age |
| ------- | ----- |
| Alice   | 30    |
| Bob     | 28    |
| Charlie | 26    |

---

## **2. ORDER BY – Sorting Results**

`ORDER BY` sorts query results by one or more properties.

**Example: Sort by age ascending**

```cypher
MATCH (p:Person)
RETURN p.name, p.age
ORDER BY p.age ASC
```

**Output:**

| p.name  | p.age |
| ------- | ----- |
| Charlie | 26    |
| Bob     | 28    |
| Alice   | 30    |

**Sort descending**

```cypher
MATCH (p:Person)
RETURN p.name, p.age
ORDER BY p.age DESC
```

**Output:**

| p.name  | p.age |
| ------- | ----- |
| Alice   | 30    |
| Bob     | 28    |
| Charlie | 26    |

---

## **3. LIMIT and SKIP – Controlling Result Size**

* **LIMIT** restricts the number of results returned
* **SKIP** skips a number of results (useful for pagination)

**Example: Top 2 oldest people**

```cypher
MATCH (p:Person)
RETURN p.name, p.age
ORDER BY p.age DESC
LIMIT 2
```

**Output:**

| p.name | p.age |
| ------ | ----- |
| Alice  | 30    |
| Bob    | 28    |

**Example: Skip the oldest**

```cypher
MATCH (p:Person)
RETURN p.name, p.age
ORDER BY p.age DESC
SKIP 1
LIMIT 2
```

**Output:**

| p.name  | p.age |
| ------- | ----- |
| Bob     | 28    |
| Charlie | 26    |

---

## **4. Aggregations – Summarizing Data**

Aggregation functions allow you to **calculate statistics or group data**.

### COUNT – Count Nodes or Relationships

**Example: Count all persons**

```cypher
MATCH (p:Person)
RETURN COUNT(p) AS total_people
```

**Output:**

| total_people |
| ------------ |
| 3            |

**Example: Count friends of Alice**

```cypher
MATCH (alice:Person {name:"Alice"})-[:FRIENDS_WITH]->(friend)
RETURN COUNT(friend) AS total_friends
```

**Output:**

| total_friends |
| ------------- |
| 2             |

---

### AVG – Average Value

**Example: Average age of all persons**

```cypher
MATCH (p:Person)
RETURN AVG(p.age) AS average_age
```

**Output:**

| average_age |
| ----------- |
| 28          |

---

### SUM – Sum of Values

**Example: Total age**

```cypher
MATCH (p:Person)
RETURN SUM(p.age) AS total_age
```

**Output:**

| total_age |
| --------- |
| 84        |

---

### COLLECT – Collecting Values into a List

**Example: List all friend names**

```cypher
MATCH (alice:Person {name:"Alice"})-[:FRIENDS_WITH]->(friend)
RETURN COLLECT(friend.name) AS friends_list
```

**Output:**

| friends_list       |
| ------------------ |
| ["Bob", "Charlie"] |

---

## **5. Grouping Data with RETURN and Aggregates**

You can **group data by a property** and then aggregate.

**Example: Count persons per city**

```cypher
MATCH (p:Person)
RETURN p.city, COUNT(p) AS people_count
```

**Output:**

| p.city  | people_count |
| ------- | ------------ |
| Colombo | 2            |
| Kandy   | 1            |

**Example: Average age per city**

```cypher
MATCH (p:Person)
RETURN p.city, AVG(p.age) AS avg_age
```

**Output:**

| p.city  | avg_age |
| ------- | ------- |
| Colombo | 28      |
| Kandy   | 28      |

---

## **6. Combining Aggregation with Filtering**

You can **filter before or after aggregation** using `WHERE` and `WITH`.

**Example: People with more than 1 friend**

```cypher
MATCH (p:Person)-[:FRIENDS_WITH]->(friend)
WITH p, COUNT(friend) AS num_friends
WHERE num_friends > 1
RETURN p.name, num_friends
```

**Output:**

| p.name | num_friends |
| ------ | ----------- |
| Alice  | 2           |

**Explanation:**

* `WITH` passes intermediate results to filter with `WHERE`
* Only Alice has more than 1 friend, so only she is returned

# **Advanced Cypher Queries**

Advanced queries in Cypher let you **traverse complex graphs, work with collections, and use built-in procedures** to handle real-world data scenarios efficiently.

---

## **1. Path Queries and Variable-Length Relationships**

Sometimes you need to **find paths of unknown length** between nodes. Cypher allows this using `*` for variable-length relationships.

### Syntax:

```
MATCH (start:Label)-[:RELATIONSHIP_TYPE*min..max]->(end:Label)
RETURN start, end
```

* `*` → variable length
* `*1..3` → 1 to 3 hops
* `*..` → unlimited hops

---

### Example 1: Find friends of friends (2 hops)

**Query:**

```cypher
MATCH (alice:Person {name:"Alice"})-[:FRIENDS_WITH*2]->(friendOfFriend)
RETURN friendOfFriend.name
```

**Output:**

| ------------------- |
| Charlie             |

**Explanation:**

* Traverses exactly 2 relationships from Alice to find “friends of friends.”

---

```cypher
MATCH path = (alice:Person {name:"Alice"})-[:FRIENDS_WITH*1..3]->(person)
RETURN path
```

**Output:**

* Returns **paths** including 1, 2, or 3 relationships away from Alice.
* `path` can be visualized in Neo4j Browser.

---

### Example 3: Path length

```cypher
MATCH path = (alice:Person {name:"Alice"})-[:FRIENDS_WITH*]->(person)
RETURN person.name, LENGTH(path) AS hops
```

**Output:**

| ----------- | ---- |
| Bob         | 1    |
| Charlie     | 1    |

# **Indexes and Constraints in Neo4j**

Indexes and constraints help **optimize performance**, **ensure data integrity**, and **enable faster search** in your graph database.

---

## **1. Performance Optimization with Indexes**

Indexes allow Neo4j to **quickly find nodes based on property values**, instead of scanning the entire graph.

### Example: Creating an Index

```cypher
CREATE INDEX person_name_index FOR (p:Person) ON (p.name)
```

**Explanation:**

* Creates an **index on the `name` property** of `Person` nodes
* Speeds up queries like:

```cypher
MATCH (p:Person {name:"Alice"})
RETURN p
```

### Listing Indexes

```cypher
SHOW INDEXES
```

**Output Table:**

| name              | type  | entityType |
| ----------------- | ----- | ---------- |
| person_name_index | BTREE | NODE       |

### Dropping an Index

```cypher
DROP INDEX person_name_index
```

---

## **2. Unique and Existence Constraints**

Constraints ensure **data integrity** in the graph.

---

### 2.1 Unique Constraint

Ensures that **no two nodes have the same property value**.

**Example: Ensure unique names**

```cypher
CREATE CONSTRAINT unique_person_name FOR (p:Person) REQUIRE p.name IS UNIQUE
```

**Behavior:**

* Prevents creation of another `Person` node with the same `name`
* Useful for **primary key-like properties**

**Test Unique Constraint**

```cypher
CREATE (p:Person {name:"Alice"})
CREATE (p:Person {name:"Alice"})   // ❌ Fails due to uniqueness constraint
```

---

### 2.2 Existence Constraint

Ensures that a property **must exist** on a node.

**Example: Ensure every Person has an age**

```cypher
CREATE CONSTRAINT person_age_exists FOR (p:Person) REQUIRE p.age IS NOT NULL
```

**Behavior:**

* Neo4j will **reject nodes** that do not have the `age` property
* Helps maintain **data consistency**

---

### Listing Constraints

```cypher
SHOW CONSTRAINTS
```

**Output Table:**

| name               | type   | entityType | properties |
| ------------------ | ------ | ---------- | ---------- |
| unique_person_name | UNIQUE | NODE       | name       |
| person_age_exists  | EXISTS | NODE       | age        |

### Dropping a Constraint

```cypher
DROP CONSTRAINT unique_person_name
```

---

## **3. Full-Text Search**

Neo4j supports **full-text indexes** to quickly search **text properties** across nodes or relationships.

---

### 3.1 Creating a Full-Text Index

```cypher
CALL db.index.fulltext.createNodeIndex("personNameIndex", ["Person"], ["name"])
```

**Explanation:**

* Index name: `personNameIndex`
* Target label: `Person`
* Properties indexed: `name`

---

### 3.2 Searching with Full-Text Index

```cypher
CALL db.index.fulltext.queryNodes("personNameIndex", "Ali*") YIELD node, score
RETURN node.name, score
```

**Output Table:**

| node.name | score |
| --------- | ----- |
| Alice     | 0.85  |

**Explanation:**

* `"Ali*"` → wildcard search (matches names starting with "Ali")
* Returns **relevant nodes with a score** indicating match quality

---

### 3.3 Dropping Full-Text Index

```cypher
CALL db.index.fulltext.drop("personNameIndex")
```

---

## **4. Practical Example Workflow**

**Step 1: Create Index and Constraint**

```cypher
CREATE INDEX person_name_index FOR (p:Person) ON (p.name)
CREATE CONSTRAINT unique_person_name FOR (p:Person) REQUIRE p.name IS UNIQUE
CREATE CONSTRAINT person_age_exists FOR (p:Person) REQUIRE p.age IS NOT NULL
```

# **Graph Data Modeling Fundamentals**

Graph data modeling is about **structuring your data as nodes, relationships, and properties** to represent real-world scenarios efficiently. A well-designed graph model improves **query performance, scalability, and clarity**.

---

## **1. Designing Entities and Relationships**

In a graph, the **two main elements** are:

1. **Nodes (Entities):** Represent objects, people, places, or concepts.

   * Example: `Person`, `Movie`, `Product`

2. **Relationships:** Represent connections or interactions between nodes.

   * Example: `FRIENDS_WITH`, `PURCHASED`, `ACTED_IN`

3. **Properties:** Key-value pairs attached to nodes or relationships for additional details.

   * Example: `Person {name: "Alice", age: 30}`
   * Example: `FRIENDS_WITH {since: 2020}`

---

### Step 1: Identify Entities

* List all main objects in your system
* Example: Social Network

  * `Person`, `Post`, `Comment`, `Group`

### Step 2: Identify Relationships

* Determine how entities interact
* Example: Social Network

  * `Person` **FRIENDS_WITH** `Person`
  * `Person` **POSTED** `Post`
  * `Person` **COMMENTED_ON** `Post`
  * `Person` **MEMBER_OF** `Group`

### Step 3: Determine Properties

* Store relevant details as properties
* Example:

  * `Person {name, age, city}`
  * `Post {content, timestamp}`
  * `FRIENDS_WITH {since}`

---

### Example Graph Representation

```
(Alice:Person {age:30})-[:FRIENDS_WITH {since:2020}]->(Bob:Person {age:28})
(Alice)-[:POSTED]->(Post1 {content:"Hello World", timestamp: "2025-10-27"})
(Bob)-[:COMMENTED_ON]->(Post1)
```

* Nodes: Alice, Bob, Post1
* Relationships: FRIENDS_WITH, POSTED, COMMENTED_ON
* Properties: age, since, content, timestamp

---

## **2. Normalization in Graph Databases**

Normalization in graphs is **different from relational databases**. The goal is to **avoid redundancy while maintaining query efficiency**.

### Principles:

1. **Avoid duplicate nodes:**

   * Each real-world entity should be represented **once**.
   * Example: Don’t create multiple `Alice` nodes; use relationships to connect her to multiple entities.

2. **Use relationships instead of repeating properties:**

   * Instead of storing a person’s friends as a property list, create **FRIENDS_WITH relationships**.

3. **Denormalize when needed for performance:**

   * Sometimes storing redundant information in nodes or relationships can **speed up queries**.
   * Example: Store `friend_count` on `Person` to quickly get number of friends.

---

### Example: Normalized Graph Model

**Bad (redundant data):**

```
Person {name:"Alice", friends:["Bob","Charlie"], posts:["Post1","Post2"]}
```

**Good (normalized):**

```
(Alice)-[:FRIENDS_WITH]->(Bob)
(Alice)-[:FRIENDS_WITH]->(Charlie)
(Alice)-[:POSTED]->(Post1)
(Alice)-[:POSTED]->(Post2)
```

**Advantages:**

* Easier to traverse relationships
* No duplication
* Queries are simpler and more efficient

---

## **3. Case Studies of Graph Models**

### Case Study 1: Social Network

**Entities:** Person, Post, Comment
**Relationships:** FRIENDS_WITH, POSTED, COMMENTED_ON
**Query Example:** Find friends of friends who commented on a post

```cypher
MATCH (alice:Person {name:"Alice"})-[:FRIENDS_WITH*2]->(fof)-[:COMMENTED_ON]->(post:Post)
RETURN fof.name, post.content
```

---

### Case Study 2: E-Commerce

**Entities:** Customer, Product, Order
**Relationships:** PURCHASED, VIEWED, REVIEWED
**Query Example:** Find products frequently purchased by friends

```cypher
MATCH (customer:Customer {name:"Alice"})-[:FRIENDS_WITH]->(friend)-[:PURCHASED]->(product)
RETURN product.name, COUNT(friend) AS friend_count
ORDER BY friend_count DESC
```

---

### Case Study 3: Knowledge Graph

**Entities:** Person, Organization, Location
**Relationships:** WORKS_AT, LIVES_IN, KNOWS
**Query Example:** Find people working at the same company

```cypher
MATCH (p1:Person)-[:WORKS_AT]->(org)<-[:WORKS_AT]-(p2:Person)
WHERE p1.name <> p2.name
RETURN p1.name, p2.name, org.name
```

---

1. **Identify all entities (nodes)**
2. **Identify all relationships between entities**
3. **Determine properties for nodes and relationships**
4. **Avoid duplicate nodes (normalize)**
5. **Add indexes and constraints for performance**

# **Best Practices for Graph Data Modeling**

Modeling graphs correctly ensures your **Neo4j database is efficient, maintainable, and scalable**. Poor design can lead to slow queries and complex maintenance.

---

## **1. Granularity of Nodes**

Granularity refers to **how detailed each node should be**. Choosing the right granularity ensures queries are **simple and efficient**.

### Guidelines:

* **One node per real-world entity**

  * Example: Each person is a node, not one node per attribute
* **Avoid overloading a node with too many responsibilities**

  * Example: Don’t combine Person + Address + Job in a single node
  * Instead:

    ```
    (Alice:Person {name:"Alice", age:30})
    (Address {street:"Main St", city:"Colombo"})
    (Job {title:"Engineer", company:"TechCorp"})
    (Alice)-[:LIVES_AT]->(Address)
    (Alice)-[:WORKS_AT]->(Job)
    ```

**Benefits of proper granularity:**

* Clear relationships
* Easier to query
* Less redundant data

---

## **2. Avoiding Over-Connected Nodes**

Over-connected nodes, also called **supernodes**, have **too many relationships**. They can slow down queries and consume memory.

### Example of Over-Connected Node:

* A node `World` connecting to **all cities, people, and organizations**
* Queries involving `World` may traverse **millions of relationships**, causing performance issues.

### Best Practices:

* **Split high-degree nodes** into smaller, meaningful nodes

  * Example: Instead of `World` → connect regions → connect countries → connect cities
* **Use intermediate nodes to reduce fan-out**

  ```
  (Region)-[:CONTAINS]->(Country)-[:HAS_CITY]->(City)
  ```

**Benefits:**

* Queries traverse fewer relationships
* Reduces memory and computation overhead

---

## **3. Handling Temporal Data**

Graphs often store **time-related information** such as events, history, or lifecycles. Proper modeling ensures **efficient querying and updates**.

### Approaches:

1. **Timestamps as Properties**

```cypher
CREATE (post:Post {content:"Hello World", created_at: datetime()})
```

* Simple for querying single events by time
* Example query: Posts after a certain date

```cypher
MATCH (p:Post)
WHERE p.created_at > datetime("2025-01-01T00:00:00")
RETURN p.content, p.created_at
```

2. **Temporal Nodes for Event History**

```cypher
(Alice:Person)-[:WORKED_AT {from: date("2020-01-01"), to: date("2022-01-01")}]->(Company)
```

* Use relationship properties to represent **time intervals**
* Enables queries like **“Where did Alice work in 2021?”**

```cypher
MATCH (a:Person {name:"Alice"})-[r:WORKED_AT]->(c:Company)
WHERE date("2021-06-01") >= r.from AND date("2021-06-01") <= r.to
RETURN c.name
```

3. **Versioned Nodes for Complex Histories**

* Create a **new node for each state change**

  ```
  (Alice_v1:Person {name:"Alice", age:30})
  (Alice_v2:Person {name:"Alice", age:31})
  ```
* Connect versions: `(Alice_v1)-[:NEXT]->(Alice_v2)`
* Useful for **audit trails or temporal analytics**

---

## **4. Practical Tips for Modeling**

1. **Use meaningful node labels and relationship types**

   * Example: `:Person`, `:Company`, `:FRIENDS_WITH`, `:WORKED_AT`

2. **Keep node properties small and relevant**

   * Don’t store large JSON blobs on nodes

3. **Avoid supernodes**

   * Use intermediate nodes for better traversal

4. **Model temporal data thoughtfully**

   * Use relationship properties or versioned nodes depending on query needs

5. **Index frequently queried properties**

   * Example: `CREATE INDEX person_name_index FOR (p:Person) ON (p.name)`

6. **Use constraints for data integrity**

   * Unique constraints: Ensure no duplicate nodes
   * Existence constraints: Ensure important properties exist

---

## **Summary Table**

| Best Practice         | Description                           | Example                                                         |
| --------------------- | ------------------------------------- | --------------------------------------------------------------- |
| Granularity           | One node per entity; don’t overload   | `Person`, `Address`, `Job` as separate nodes                    |
| Avoid Supernodes      | Split high-degree nodes               | `Region->Country->City` instead of `World->AllCities`           |
| Temporal Data         | Store time efficiently                | Relationship property `WORKED_AT {from, to}` or versioned nodes |
| Meaningful Labels     | Clear node and relationship types     | `:Person`, `:FRIENDS_WITH`                                      |

# **Schema Design Examples in Graph Databases**

Graph schema design involves **deciding the nodes, relationships, and properties** for your application. Let’s explore **common use cases**.

---

## **1. Social Networks**

### Requirements:

* Users can **befriend** each other
* Users can **post content**
* Users can **like or comment** on posts

### Nodes:

* `Person` (name, age, location)
* `Post` (content, timestamp)

### Relationships:

* `FRIENDS_WITH` between `Person` nodes
* `POSTED` between `Person` → `Post`
* `LIKED` between `Person` → `Post`
* `COMMENTED_ON` between `Person` → `Post`

### Example Schema:

```
(Alice:Person {name:"Alice"})-[:FRIENDS_WITH]->(Bob:Person {name:"Bob"})
(Alice)-[:POSTED]->(Post1 {content:"Hello World", timestamp:"2025-10-27"})
(Bob)-[:LIKED]->(Post1)
(Bob)-[:COMMENTED_ON]->(Post1)
```

### Sample Queries:

* Find friends of Alice:

```cypher
MATCH (alice:Person {name:"Alice"})-[:FRIENDS_WITH]->(friend)
RETURN friend.name
```

* Find posts liked by friends:

```cypher
MATCH (alice:Person {name:"Alice"})-[:FRIENDS_WITH]->(friend)-[:LIKED]->(post)
RETURN friend.name, post.content
```

---

## **2. Knowledge Graphs**

### Requirements:

* Represent **entities** and **relationships** between them
* Support complex queries and semantic relationships

### Nodes:

* `Person` (name, occupation)
* `Organization` (name, type)
* `Location` (name, type)

### Relationships:

* `WORKS_AT` (Person → Organization)
* `LIVES_IN` (Person → Location)
* `KNOWS` (Person → Person)

### Example Schema:

```
(Alice:Person)-[:WORKS_AT]->(TechCorp:Organization)
(Alice)-[:LIVES_IN]->(Colombo:Location)
(Alice)-[:KNOWS]->(Bob:Person)
```

### Sample Queries:

* Find people working at the same organization:

```cypher
MATCH (p1:Person)-[:WORKS_AT]->(org)<-[:WORKS_AT]-(p2:Person)
WHERE p1.name <> p2.name
RETURN p1.name, p2.name, org.name
```

* Find connections within 2 hops:

```cypher
MATCH (alice:Person {name:"Alice"})-[:KNOWS*1..2]->(connection)
RETURN connection.name
```

---

## **3. Recommendation Engines**

### Requirements:

* Recommend products to users based on **past purchases, views, or ratings**
* Capture relationships between users and products

### Nodes:

* `User` (name, age, preferences)
* `Product` (name, category, price)

### Relationships:

* `PURCHASED` (User → Product)
* `VIEWED` (User → Product)
* `RATED` (User → Product {rating:1-5})

### Example Schema:

```
(Alice:User)-[:PURCHASED]->(Laptop:Product {name:"Laptop", category:"Electronics"})
(Bob:User)-[:PURCHASED]->(Laptop)
(Alice)-[:VIEWED]->(Headphones:Product {name:"Headphones"})
(Bob)-[:RATED {rating:5}]->(Laptop)
```

### Sample Queries:

* Recommend products bought by friends:

```cypher
MATCH (user:User {name:"Alice"})-[:FRIENDS_WITH]->(friend)-[:PURCHASED]->(product)
RETURN product.name, COUNT(friend) AS popularity
ORDER BY popularity DESC
```

* Find products similar to purchased items (same category):

```cypher
MATCH (alice:User {name:"Alice"})-[:PURCHASED]->(p:Product)
MATCH (rec:Product {category: p.category})
WHERE rec <> p
RETURN rec.name
```

## **4. Practical Tips for Schema Design**

1. **Identify the core entities**: Users, products, organizations, posts
2. **Define relationships clearly**: Make relationship types descriptive (`FRIENDS_WITH`, `PURCHASED`)
3. **Use properties wisely**: Store frequently queried data as node/relationship properties
4. **Avoid supernodes**: Split high-degree nodes into intermediate nodes
5. **Model for queries**: Design your graph **around the queries you will run most often**
6. **Use indexes and constraints**: Improve performance and data integrity

# **Introduction to Graph Data Science (GDS)**

Graph Data Science (GDS) is a set of **algorithms and tools for analyzing graph data**. It enables organizations to uncover patterns, identify important nodes, and make predictions using the relationships in their graph.

Neo4j provides the **GDS library**, which is optimized for **performance and scalability**, allowing you to run algorithms directly on large graphs efficiently.

---

## **1. Overview of the GDS Library**

**Key Features:**

* Supports **graph analytics** for **centrality, similarity, community detection, and pathfinding**
* Works with **projected graphs** in memory for fast computation
* Integrates with **APOC, Cypher, and Neo4j GraphQL**

**Installation:**

* GDS is a plugin for Neo4j Enterprise Edition
* Enable in `neo4j.conf`:

```text
dbms.security.procedures.unrestricted=gds.*
```

* Restart Neo4j after installation

**Common Use Cases:**

* Detecting influential users in social networks
* Recommending products or content
* Fraud detection in financial networks
* Network optimization

---

## **2. Graph Projections and Pipelines**

Before running algorithms, GDS requires a **graph projection**. This is an **in-memory subgraph** optimized for computation.

### Creating a Graph Projection

```cypher
CALL gds.graph.create(
  'socialGraph',           // name of the projected graph
  'Person',                // node labels
  'FRIENDS_WITH'           // relationship types
)
```

* **Node labels:** Select which nodes to include
* **Relationship types:** Choose which relationships are considered
* Graph is now **stored in memory**, enabling fast analytics

### Pipelines

* **Graph pipelines** combine **projection, algorithm execution, and writing results back**
* Example:

```cypher
CALL gds.pageRank.stream('socialGraph')
YIELD nodeId, score
RETURN gds.util.asNode(nodeId).name AS name, score
ORDER BY score DESC
```

---

## **3. Centrality Algorithms**

Centrality measures **importance or influence of nodes** in the network. Neo4j GDS provides several centrality algorithms.

---

### 3.1 PageRank

* Measures the **importance of nodes** based on their connections
* Originally used by Google for ranking websites

**Example Query:**

```cypher
CALL gds.pageRank.stream('socialGraph')
YIELD nodeId, score
RETURN gds.util.asNode(nodeId).name AS name, score
ORDER BY score DESC
```

**Output Example:**

| name    | score |
| ------- | ----- |
| Alice   | 0.34  |
| Bob     | 0.28  |
| Charlie | 0.22  |

**Explanation:**

* Nodes with many incoming connections or connections from important nodes have higher scores

---

### 3.2 Betweenness Centrality

* Measures the **number of shortest paths that pass through a node**
* Identifies nodes that act as **bridges** in the network

**Example Query:**

```cypher
CALL gds.betweenness.stream('socialGraph')
YIELD nodeId, score
RETURN gds.util.asNode(nodeId).name AS name, score
ORDER BY score DESC
```

**Use Case:**

* Find **key influencers** who connect different communities

---

### 3.3 Closeness Centrality

* Measures how **close a node is to all other nodes** in the graph
* Nodes with high closeness can **quickly reach the rest of the network**

**Example Query:**

```cypher
CALL gds.closeness.stream('socialGraph')
YIELD nodeId, score
RETURN gds.util.asNode(nodeId).name AS name, score
ORDER BY score DESC
```

**Use Case:**

* Identify nodes **best positioned for spreading information or alerts**

---

# **Pathfinding and Similarity Algorithms**

Graph algorithms help **analyze relationships, traverse networks, and find patterns**. Neo4j’s Graph Data Science (GDS) library provides **pathfinding and similarity algorithms** to solve these problems efficiently.

---

## **1. Pathfinding Algorithms**

Pathfinding algorithms find **routes or paths between nodes**, often optimizing for distance, cost, or other weights.

---

### 1.1 Shortest Path

* Finds the **minimum number of hops** between two nodes
* Useful when **edges are unweighted**

**Example Query:**

```cypher
MATCH (alice:Person {name:"Alice"}), (bob:Person {name:"Bob"})
CALL gds.shortestPath.stream({
  nodeProjection: 'Person',
  relationshipProjection: 'FRIENDS_WITH',
  startNode: alice,
  endNode: bob
})
YIELD nodeIds, costs
RETURN [nodeId IN nodeIds | gds.util.asNode(nodeId).name] AS path, costs
```

**Output Example:**

| path                  | costs |
| --------------------- | ----- |
| [Alice, Charlie, Bob] | 2     |

**Explanation:**

* Finds the path with **least number of relationships** connecting Alice to Bob

---

### 1.2 Dijkstra Algorithm

* Computes **shortest paths on weighted graphs**
* Edge weights can represent **distance, cost, or time**

**Example Query:**

```cypher
CALL gds.shortestPath.dijkstra.stream({
  nodeProjection: 'City',
  relationshipProjection: {
    ROAD: {
      type: 'ROAD',
      properties: 'distance'
    }
  },
  startNode: gds.util.asNode('City1'),
  endNode: gds.util.asNode('City5')
})
YIELD nodeIds, totalCost
RETURN [nodeId IN nodeIds | gds.util.asNode(nodeId).name] AS path, totalCost
```

**Use Case:**

* Road networks, transportation planning, logistics routing

---

### 1.3 A* (A-Star) Algorithm

* Optimized pathfinding using **heuristics**
* Finds the shortest path **faster than Dijkstra** in large graphs
* Requires a **heuristic function**, often **Euclidean distance or straight-line distance**

**Example Query:**

```cypher
CALL gds.alpha.shortestPath.astar.stream({
  nodeProjection: 'City',
  relationshipProjection: 'ROAD',
  startNode: gds.util.asNode('City1'),
  endNode: gds.util.asNode('City5'),
  relationshipWeightProperty: 'distance',
  latitudeProperty: 'lat',
  longitudeProperty: 'lon'
})
YIELD nodeIds, totalCost
RETURN [nodeId IN nodeIds | gds.util.asNode(nodeId).name] AS path, totalCost
```

**Use Case:**

* Navigation apps, route optimization, game AI

---

## **2. Node Similarity Algorithms**

Node similarity algorithms **measure how similar two nodes are**, based on shared neighbors or properties. This is useful for **recommendations, clustering, and community detection**.

---

### 2.1 Jaccard Coefficient

* Measures similarity as:
  [
  \text{Jaccard}(A, B) = \frac{|N(A) \cap N(B)|}{|N(A) \cup N(B)|}
  ]
  Where (N(A)) = neighbors of node A

**Example Query:**

```cypher
CALL gds.nodeSimilarity.stream({
  nodeProjection: 'Person',
  relationshipProjection: 'FRIENDS_WITH'
})
YIELD node1, node2, similarity
RETURN gds.util.asNode(node1).name AS person1,
       gds.util.asNode(node2).name AS person2,
       similarity
ORDER BY similarity DESC
```

**Output Example:**

| person1 | person2 | similarity |
| ------- | ------- | ---------- |
| Alice   | Charlie | 0.75       |
| Alice   | Bob     | 0.5        |

**Explanation:**

* Alice and Charlie share **75% of their friends**, so they are highly similar
* Useful for **friend recommendations or grouping similar nodes**

---

### 2.2 Use Cases of Node Similarity

* **Social networks:** Suggest friends with common connections
* **E-commerce:** Recommend products based on similar user purchases
* **Knowledge graphs:** Find related entities based on shared relationships

---

## 3. Practical Tips

1. **Use weighted pathfinding** (Dijkstra/A*) when relationships have meaningful costs
2. **A* is faster than Dijkstra** for large graphs if a good heuristic exists
3. **Node similarity algorithms** are helpful for **recommendations and clustering**
4. Combine **pathfinding and similarity** for advanced analytics:

   * Example: Find shortest path between similar nodes

---

## **Summary Table**

| Algorithm           | Purpose                       | Use Case                          | Notes                                    |
| ------------------- | ----------------------------- | --------------------------------- | ---------------------------------------- |
| Shortest Path       | Fewest hops                   | Social network connections        | Unweighted graphs                        |
| Dijkstra            | Weighted shortest path        | Road networks, logistics          | Uses edge weights                        |
| A*                  | Heuristic-based shortest path | Navigation, game AI               | Faster than Dijkstra with good heuristic |
| Jaccard Coefficient | Node similarity               | Friend or product recommendations | Based on shared neighbors                |


# **Performance Tuning and Security in Neo4j**

---

## **1. Query Profiling and Optimization**

Query performance is key for **large graphs**.

### Query Profiling

* Use the **`PROFILE`** or **`EXPLAIN`** commands to analyze query execution:

```cypher
EXPLAIN MATCH (p:Person)-[:FRIENDS_WITH]->(f)
RETURN p.name, f.name
```

* **EXPLAIN**: Shows query plan without execution
* **PROFILE**: Executes query and shows detailed statistics

### Tips for Optimization

1. **Use indexes and constraints**

   ```cypher
   CREATE INDEX person_name_index FOR (p:Person) ON (p.name)
   CREATE CONSTRAINT unique_person_name IF NOT EXISTS FOR (p:Person) REQUIRE p.name IS UNIQUE
   ```

   * Improves lookup speed for `MATCH` queries

2. **Avoid unnecessary MATCH clauses**

   * Match only nodes needed for the query

3. **Use `WITH` to limit intermediate results**

```cypher
MATCH (p:Person)-[:FRIENDS_WITH]->(f)
WITH p, f LIMIT 100
RETURN p.name, f.name
```

4. **Prefer `MERGE` carefully**

   * `MERGE` checks for existence before creation, can be slower than `CREATE`

5. **Profile long-running queries** to identify **expensive operations** like Cartesian products

## **2. Caching and Memory Configuration**

### Memory Configuration

* Set in `neo4j.conf`:

```text
dbms.memory.heap.initial_size=4G
dbms.memory.heap.max_size=8G
dbms.memory.pagecache.size=6G
```

**Tips:**

* Heap memory: Used for queries and transactions
* Page cache: Stores graph data for **fast access**

### Caching Strategies

* Frequently accessed nodes and relationships are **cached in memory**

## **3. Transaction Management**

Transactions ensure **data consistency** in Neo4j.

### Basics

* **COMMIT** saves changes

# **Neo4j and Machine Learning**

Graphs are **rich sources of relational information**, which can enhance ML models. Neo4j allows you to **extract graph features** and **generate embeddings** for downstream ML tasks such as node classification, link prediction, or recommendations.

## **1. Graph Embeddings and Feature Extraction**

### What are Graph Embeddings?

* **Graph embeddings** map nodes, edges, or entire subgraphs into **low-dimensional vectors**
* Preserve **structural and semantic information**
* Enable ML algorithms (which usually work on tabular data) to use graph information

### Common Techniques:

1. **Node2Vec**

   * Generates embeddings by simulating random walks on the graph
   * Captures neighborhood structures

2. **GraphSAGE**

   * Aggregates information from a node’s neighbors
   * Suitable for inductive learning

3. **FastRP (Fast Random Projection)**

   * Neo4j’s built-in algorithm for fast node embeddings
   * Scales well to large graphs

---

### Example: Using FastRP in Neo4j GDS

```cypher
CALL gds.fastRP.write({
  nodeProjection: 'Person',
  relationshipProjection: 'FRIENDS_WITH',
  embeddingDimension: 64,
  writeProperty: 'embedding'
})
YIELD nodePropertiesWritten
```

* Stores **64-dimensional embeddings** in the `embedding` property of each node
* Can be used directly in ML pipelines

**Example Output:**

| name  | embedding (vector)              |
| ----- | ------------------------------- |
| Alice | [0.12, -0.34, 0.56, ..., 0.01]  |
| Bob   | [-0.21, 0.44, -0.11, ..., 0.33] |

---

### Feature Extraction

* Extract **graph-based features** such as:

  * **Degree centrality**
  * **PageRank scores**
  * **Clustering coefficients**
* Example: Add centrality features to nodes for ML:

```cypher
CALL gds.pageRank.write({
  nodeProjection: 'Person',
  relationshipProjection: 'FRIENDS_WITH',
  writeProperty: 'pagerank'
})
```

* Now each node has `pagerank` as a feature for ML models

## **2. Using Neo4j with PyTorch Geometric**

**PyTorch Geometric (PyG)** is a library for **graph neural networks (GNNs)**. You can combine Neo4j and PyG for **deep learning on graph data**.

### Steps:

1. **Extract graph data from Neo4j**

```python
from neo4j import GraphDatabase
import torch
from torch_geometric.data import Data

driver = GraphDatabase.driver("bolt://localhost:7687", auth=("neo4j", "password"))

with driver.session() as session:
    result = session.run("MATCH (p:Person) RETURN p.name AS name, p.embedding AS emb")
    embeddings = [record["emb"] for record in result]

    result = session.run("MATCH (p1:Person)-[:FRIENDS_WITH]->(p2:Person) RETURN p1.name AS src, p2.name AS dst")
    edges = [(record["src"], record["dst"]) for record in result]
```

2. **Convert data to PyG format**

```python
import torch

# Example: mapping names to indices
node_map = {name: idx for idx, name in enumerate([record["name"] for record in result])}
edge_index = torch.tensor([[node_map[src], node_map[dst]] for src, dst in edges], dtype=torch.long).t().contiguous()

x = torch.tensor(embeddings, dtype=torch.float)

data = Data(x=x, edge_index=edge_index)
```

3. **Use GNN models**

```python
from torch_geometric.nn import GCNConv

class GCN(torch.nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()
        self.conv1 = GCNConv(in_channels, 16)
        self.conv2 = GCNConv(16, out_channels)
    def forward(self, data):
        x, edge_index = data.x, data.edge_index
        x = self.conv1(x, edge_index).relu()
        x = self.conv2(x, edge_index)
        return x
```

* This allows **node classification, link prediction, and clustering**

# **Building RAG Systems with Neo4j**

**RAG systems** combine **retrieval-based knowledge** with **generative models**. Neo4j can act as a **retriever** storing structured and vectorized knowledge, which can then be augmented by LLMs like OpenAI’s GPT models.

---

## **1. Using Neo4j as a Retriever for Vectorized Knowledge**

### Vector Storage in Neo4j

* Neo4j supports storing **embeddings** as **arrays of floats** on nodes or relationships:

```cypher
CREATE (p:Document {
    id: "doc1",
    text: "Neo4j is a graph database",
    embedding: [0.12, 0.34, ..., 0.98]
})
```

* Use **GDS** or external libraries (like OpenAI embeddings) to generate vectors
* Index embeddings for **efficient similarity search**:

```cypher
CALL gds.knn.createNodeIndex({
  nodeProjection: 'Document',
  embeddingProperty: 'embedding'
})
```

### Vector Similarity Retrieval

* Neo4j GDS provides **k-nearest neighbors (KNN)** search:

```cypher
CALL gds.knn.stream({
  nodeProjection: 'Document',
  embeddingProperty: 'embedding',
  topK: 3,
  vector: [0.1, 0.2, ..., 0.9]
})
YIELD nodeId, similarity
RETURN gds.util.asNode(nodeId).text AS doc, similarity
ORDER BY similarity DESC
```

* Retrieves the **top-K most relevant documents** based on vector similarity

## **2. Integrating with LangChain / LangGraph**

### LangChain

* Framework for building **LLM-based pipelines**
* Steps for Neo4j RAG:

  1. **Embed documents** (OpenAI embeddings)
  2. **Store embeddings in Neo4j**
  3. **Retrieve top-K similar documents**
  4. **Use retrieved docs as context** for LLM generation

**Python Example with LangChain:**

```python
from langchain.vectorstores import Neo4j
from langchain.embeddings.openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings()
vectorstore = Neo4j(
    url="bolt://localhost:7687",
    username="neo4j",
    password="password",
    embedding=embeddings
)

query = "What is Neo4j?"
docs = vectorstore.similarity_search(query, k=3)
```

### LangGraph

* LangGraph provides **agentic and pipeline-oriented workflows**
* You can define nodes for **retriever**, **LLM generator**, and **re-ranker**
* Neo4j serves as the **retrieval node**, feeding relevant context to LLMs

**Example Pipeline:**

```
User Query → Neo4j Retriever → Top-K Docs → LLM → Answer
```

## **3. Example: RAG with Neo4j + OpenAI**

### Step 1: Embed Documents

```python
from openai import OpenAI
import numpy as np

client = OpenAI(api_key="YOUR_API_KEY")
documents = [
    {"id": "doc1", "text": "Neo4j is a graph database."},
    {"id": "doc2", "text": "RAG systems combine retrieval with generation."}
]

for doc in documents:
    response = client.embeddings.create(
        input=doc["text"],
        model="text-embedding-3-small"
    )
    doc["embedding"] = response.data[0].embedding
```

### Step 2: Store in Neo4j

```cypher
UNWIND $docs AS doc
CREATE (d:Document {id: doc.id, text: doc.text, embedding: doc.embedding})
```

### Step 3: Retrieve Similar Documents

```python
query_embedding = client.embeddings.create(
    input="Explain Neo4j RAG systems",
    model="text-embedding-3-small"
).data[0].embedding

similar_docs = vectorstore.similarity_search_by_vector(query_embedding, k=2)
```

### Step 4: Generate Answer with LLM

```python
from langchain.chat_models import ChatOpenAI
from langchain.chains import RetrievalQA

llm = ChatOpenAI(model="gpt-4o")
qa = RetrievalQA.from_chain_type(llm=llm, retriever=vectorstore.as_retriever())
answer = qa.run("Explain Neo4j RAG systems")
print(answer)
```

**Flow Explained:**

1. Query is embedded into a vector
2. Neo4j retrieves top-K similar documents
3. LLM uses the retrieved documents as **context** to generate accurate answers

# **Scaling Neo4j in Production**

Neo4j is designed to handle **large-scale graph workloads**, but production systems require **scaling, reliability, and resilience**. This includes clustering, high availability, backup strategies, and cloud deployment.

---

## **1. Clustering and High Availability**

### What is Clustering?

* Clustering allows multiple Neo4j instances to work together
* Provides **fault tolerance, high availability, and read scalability**

### Neo4j Cluster Types

1. **Causal Clustering** (Enterprise Edition)

   * Nodes have roles: **Leader, Followers, Read Replicas**
   * Leader handles **writes**
   * Followers handle **read requests**
   * Read replicas increase **read throughput**

**Roles Explained:**

| Role         | Responsibility                                             |
| ------------ | ---------------------------------------------------------- |
| Leader       | Accepts writes, coordinates replication                    |
| Follower     | Replicates data from leader, handles reads                 |
| Read Replica | Optional node for read scaling, does not vote on consensus |

### Setting Up a Causal Cluster

* Minimum 3 core nodes recommended
* Example `neo4j.conf` for a core node:

```text
dbms.mode=CORE
causal_clustering.discovery_list=neo4j-core1:5000,neo4j-core2:5000,neo4j-core3:5000
```

* Start all nodes, they elect a **leader automatically**

**Benefits:**

* Continuous availability during node failures
* Horizontal read scaling
* Strong consistency for writes

---

## **2. Backup and Recovery**

### Backup Types

1. **Online Backup (Hot Backup)**

   * Enterprise Edition only
   * Backup while the database is running
   * Command:

```bash
neo4j-admin backup --backup-dir=/backups --name=my_backup
```

2. **Offline Backup (Cold Backup)**

   * Shutdown Neo4j before copying the database
   * Simpler but **downtime required**

### Recovery

* Restore a backup:

```bash
neo4j-admin restore --from=/backups/my_backup --database=neo4j --force
```

### Best Practices

* **Automate backups** daily or hourly depending on RTO/RPO
* **Store backups offsite** or in cloud storage
* **Test recovery regularly** to ensure reliability

# **Recommendation System with Neo4j**

Recommendation systems suggest **items, users, or content** based on **user behavior, content similarity, or graph relationships**. In Neo4j, graph algorithms and embeddings make recommendations **fast and insightful**.

---

## step 1: Graph Data Modeling

### Nodes

* `User` → properties: `name`, `email`
* `Paper` → properties: `title`, `abstract`, `embedding`
* `Topic` → properties: `name`

### Relationships

* `:READ` → User → Paper
* `:LIKES` → User → Paper
* `:HAS_TOPIC` → Paper → Topic

**Example Data:**

```cypher
CREATE (alice:User {name:"Alice"})
CREATE (bob:User {name:"Bob"})
CREATE (paper1:Paper {title:"Graph Databases 101"})
CREATE (paper2:Paper {title:"Introduction to Neo4j"})
CREATE (topic:Topic {name:"Graph Databases"})
CREATE (alice)-[:READ]->(paper1)
CREATE (bob)-[:READ]->(paper1)
CREATE (paper1)-[:HAS_TOPIC]->(topic)
CREATE (paper2)-[:HAS_TOPIC]->(topic)
```

---

## Step 2: Recommendation Approaches

Neo4j provides **multiple algorithms** for recommendations:

### 2.1 Similarity-Based Recommendations

* Based on **node similarity** (common neighbors or embeddings)

```cypher
CALL gds.nodeSimilarity.stream({
  nodeProjection: 'Paper',
  relationshipProjection: 'HAS_TOPIC',
  similarityCutoff: 0.5
})
YIELD node1, node2, similarity
RETURN gds.util.asNode(node1).title AS paper1,
       gds.util.asNode(node2).title AS paper2,
       similarity
ORDER BY similarity DESC
```

**Use Case:** Suggest papers **similar to ones the user has read**

---

### 2.2 Collaborative Filtering

* Use **user behavior** (e.g., READ, LIKES) to find similar users and recommend items

```cypher
MATCH (u1:User {name:"Alice"})-[:READ]->(p:Paper)<-[:READ]-(u2:User)
WITH u1, u2, COUNT(p) AS common
MATCH (u2)-[:READ]->(rec:Paper)
WHERE NOT (u1)-[:READ]->(rec)
RETURN rec.title AS recommendation, common
ORDER BY common DESC
```

**Explanation:**

* Find users with overlapping readings
* Recommend papers they read that Alice hasn’t seen

---

### 2.3 Graph Embedding-Based Recommendations

* Use **GDS algorithms** like **FastRP** or **Node2Vec** to generate embeddings

```cypher
CALL gds.fastRP.write({
  nodeProjection: 'Paper',
  relationshipProjection: 'HAS_TOPIC',
  embeddingDimension: 64,
  writeProperty: 'embedding'
})
YIELD nodePropertiesWritten
```

* Retrieve **top-K recommendations** using **cosine similarity**:

```cypher
CALL gds.knn.stream({
  nodeProjection: 'Paper',
  embeddingProperty: 'embedding',
  topK: 3,
  nodeId: $paperId
})
YIELD nodeId, similarity
RETURN gds.util.asNode(nodeId).title AS recommendedPaper, similarity
```

---

## Step 3: Real-Time Recommendation Workflow

1. User reads or likes a paper
2. System retrieves **similar papers or papers liked by similar users**
3. Use **embeddings or node similarity** to rank recommendations
4. Return **top-N recommendations** to the user

---

## Step 4: Integrating with Python / ML

### Using Neo4j with Python

```python
from neo4j import GraphDatabase

driver = GraphDatabase.driver("bolt://localhost:7687", auth=("neo4j", "password"))

def recommend_papers(user_name, top_k=5):
    with driver.session() as session:
        result = session.run("""
        MATCH (u:User {name:$user_name})-[:READ]->(p:Paper)<-[:READ]-(other:User)
        WITH u, other, COUNT(p) AS common
        MATCH (other)-[:READ]->(rec:Paper)
        WHERE NOT (u)-[:READ]->(rec)
        RETURN rec.title AS recommendation, common
        ORDER BY common DESC
        LIMIT $top_k
        """, user_name=user_name, top_k=top_k)
        return [record["recommendation"] for record in result]

print(recommend_papers("Alice"))
```

**Output Example:**

```
['Introduction to Neo4j', 'Graph Analytics Basics']
```
